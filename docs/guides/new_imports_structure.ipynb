## Import Structure Changes

The `probatus` package has undergone a reorganization of its internal structure. The `__init__.py` file in the package contains code that **maintains backward compatibility**, allowing users to import using old paths while the package internally uses a new structure.

## The Package Structure

Probatus is organized into several subpackages:

- `probatus.core`: Base classes
- `probatus.dataset`: Dataset analysis tools including resemblance modeling
- `probatus.features`: Feature selection and importance tools
- `probatus.model`: Model interpretation tools
- `probatus.utils`: Utility functions

## Demonstrating the Old Import Style

The legacy import paths still work due to the backward compatibility code in `__init__.py`:

```python
# Old (legacy) import style
from probatus.sample_similarity import BaseResemblanceModel
from probatus.interpret import ShapModelInterpreter
from probatus.feature_elimination import ShapRFECV

print(f"BaseResemblanceModel from legacy import: {BaseResemblanceModel.__module__}")
print(f"ShapModelInterpreter from legacy import: {ShapModelInterpreter.__module__}")
print(f"ShapRFECV from legacy import: {ShapRFECV.__module__}")
```

## Demonstrating the New Import Style

The new import paths reflect the reorganized package structure:

```python
# New import style
from probatus.dataset.resemblance_modeler import BaseResemblanceModel as NewBaseResemblanceModel
from probatus.model.shap_interpreter import ShapModelInterpreter as NewShapModelInterpreter
from probatus.features.shap_recursive_feature_elimination import ShapRFECV as NewShapRFECV

print(f"BaseResemblanceModel from new import: {NewBaseResemblanceModel.__module__}")
print(f"ShapModelInterpreter from new import: {NewShapModelInterpreter.__module__}")
print(f"ShapRFECV from new import: {NewShapRFECV.__module__}")
```

## Conclusion

While both import styles currently work, it's recommended to use the new import paths. The `probatus` package's `__init__.py` file provides backward compatibility while allowing for a more organized internal structure. As a user of the package, you should prefer the new import paths for future-proof code, but existing code using the old import paths will continue to work for now.
